```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
from rapidfuzz import process, fuzz
import gradio as gr

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

def match_user_input(user_input, choices, cutoff=60):
    if not user_input:
        return None
    user_input = user_input.strip().title()  # Standardize user input
    match = process.extractOne(user_input, choices, scorer=fuzz.WRatio, score_cutoff=cutoff)
    return match[0] if match else None

def apply_make_exceptions(user_input):
    make_exceptions = {
        'BMW': 'BMW', 'bmw': 'BMW', 'Bmw': 'BMW',
        'GMC': 'GMC', 'Gmc': 'GMC', 'gmc': 'GMC', 'MINI': 'MINI', 'Mini': 'MINI', 'mini': 'MINI',
        'VW': 'Volkswagen', 'vw': 'Volkswagen', 'Vw': 'Volkswagen', 'VOLKSWAGEN': 'Volkswagen',
        'Tofas': 'Tofaş', 'tofas': 'Tofaş'
    }
    return make_exceptions.get(user_input, None)

def calculate_engine_power(make, model, year, engine_size):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'].between(year - 1, year + 1)) & 
        (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
    ]
    
    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
        
    if len(filtered_df) == 0:
        filtered_df = df_2024[
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
    
    if len(filtered_df) == 0:
        filtered_df = df_2024[df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05)]
    
    engine_power = filtered_df['Engine Power (hp)'].median()
    if pd.isna(engine_power):
        engine_power = 75  # Default placeholder if no data available
    return engine_power

def calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Fuel Type'] == fuel_type) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
    ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 2, year + 2)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        for i in range(1, 5):
            filtered_df = df_2024[
                (df_2024['Make'] == make) &
                (df_2024['Model'] == model) &
                (df_2024['Year'].between(year - i, year + i)) &
                (df_2024['Kilometers (km)'].between(km - 15000 * (i + 1), km + 15000 * (i + 1)))
            ]
            if len(filtered_df) >= 10:
                break

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) &
            (df_2024['Model'] == model)
        ]

    if len(filtered_df) < 10:
        base_confidence = "Low"
    elif len(filtered_df) < 30:
        base_confidence = "Medium"
    else:
        base_confidence = "High"

    if len(filtered_df) >= 10:
        filtered_df = filtered_df.sort_values(by='Price USD')
        base_line_price = filtered_df['Price USD'].median()
    else:
        base_line_price = filtered_df['Price USD'].median() if len(filtered_df) > 0 else overall_median_price

    base_std_dev = filtered_df['Price USD'].std()

    return base_line_price, base_confidence, filtered_df, base_std_dev

def calculate_weights(base_confidence, model_confidence, agreement_level):
    if base_confidence == "High" and model_confidence == "High":
        if agreement_level == "High":
            return 0.6, 0.4
        elif agreement_level == "Medium":
            return 0.65, 0.35
        else:
            return 0.7, 0.3
    elif base_confidence == "High" and model_confidence == "Low":
        if agreement_level == "High":
            return 0.7, 0.3
        elif agreement_level == "Medium":
            return 0.75, 0.25
        else:
            return 0.8, 0.2
    elif base_confidence == "Low" and model_confidence == "High":
        if agreement_level == "High":
            return 0.5, 0.5
        elif agreement_level == "Medium":
            return 0.55, 0.45
        else:
            return 0.6, 0.4
    else:
        if agreement_level == "High":
            return 0.55, 0.45
        elif agreement_level == "Medium":
            return 0.6, 0.4
        else:
            return 0.65, 0.35

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()
    make_model = f"{make}_{model}".replace(' ', '_')

    model_path = f"MakeModel/{make_model}_model.pkl"
    scaler_path = f"MakeModel/{make_model}_scaler.pkl"
    encoder_path = "MakeModel/onehot_encoder.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return f"Model not available for {make_model}. Unable to estimate price."
    
    car_age = datetime.now().year - year
    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(
        make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type
    )

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)

    # Load the encoder and scaler
    encoder = joblib.load(encoder_path)
    scaler = joblib.load(scaler_path)

    # Specify the numerical and categorical columns
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    # Required columns by the model
    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    # Encode the categorical features
    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))

    # Concatenate the numerical features and encoded categorical features
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    # Ensure all required columns are present
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    # Reorder the columns to match the model's expected input
    user_data_final = user_data_final[required_columns]

    # Scale the user data
    user_data_scaled = scaler.transform(user_data_final)

    # Load the model
    model = joblib.load(model_path)

    # Predict the price using the model
    model_predicted_price = model.predict(user_data_scaled)[0]

    # Calculate agreement level
    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    # Calculate confidence in the model's prediction using test max error percentage and value count
    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    # Determine weights based on confidences and agreement
    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)

    # Calculate the final price estimate
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    # Prepare results to return
    results = f"""
    Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, {traction} traction: {final_price:.2f} USD
    Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})
    Model confidence level: {model_confidence}
    Agreement level: {agreement_level}
    Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%
    Filtered data used for base-line estimate:
    {base_filtered_data[['Price USD', 'Year', 'Kilometers (km)', 'Engine Size (cc)', 'Body Type', 'Transmission']]}
    Input data for model prediction (non-zero values):
    {user_data_final.loc[:, (user_data_final != 0).any(axis=0)]}
    """
    return results

# Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("# Car Price Estimator")
    
    with gr.Row():
        make = gr.Textbox(label="Make")
        model = gr.Textbox(label="Model")
    
    with gr.Row():
        year = gr.Number(label="Year", precision=0)
        engine_size = gr.Number(label="Engine Size (cc)", precision=0)
        km = gr.Number(label="Kilometers", precision=0)
    
    with gr.Row():
        color = gr.Textbox(label="Color")
        transmission = gr.Textbox(label="Transmission")
        body_type = gr.Textbox(label="Body Type")
        fuel_type = gr.Textbox(label="Fuel Type")
        traction = gr.Textbox(label="Traction")
    
    estimate_button = gr.Button("Estimate Price")
    
    result = gr.Textbox(label="Estimated Price")
    
    estimate_button.click(
        estimate_car_value, 
        inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction],
        outputs=result
    )

demo.launch()
```

```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
import gradio as gr
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

makes = sorted(df_2024['Make'].unique())
colors = sorted(df_2024['Color'].unique())

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.Dropdown.update(choices=models)

def update_transmissions(make, model, year):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year)
    ]
    transmissions = filtered_df['Transmission'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=transmissions)

def update_body_types(make, model, year, transmission):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission)
    ]
    body_types = filtered_df['Body Type'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=body_types)

def update_fuel_types(make, model, year, transmission, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type)
    ]
    fuel_types = filtered_df['Fuel Type'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=fuel_types)

def update_tractions(make, model, year, transmission, body_type, fuel_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Fuel Type'] == fuel_type)
    ]
    tractions = filtered_df['Traction'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=tractions)

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    # Load the necessary models and encoders
    make_model = f"{make}_{model}".replace(' ', '_')
    model_path = f"MakeModel/{make_model}_model.pkl"
    scaler_path = f"MakeModel/{make_model}_scaler.pkl"
    encoder_path = "MakeModel/onehot_encoder.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return f"Model not available for {make_model}. Unable to estimate price."
    
    model = joblib.load(model_path)
    scaler = joblib.load(scaler_path)
    encoder = joblib.load(encoder_path)

    car_age = datetime.now().year - year

    def calculate_engine_power(make, model, year, engine_size):
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
        
        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) & 
                (df_2024['Model'] == model) & 
                (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
            ]
            
        if len(filtered_df) == 0:
            filtered_df = df_2024[
                (df_2024['Year'].between(year - 1, year + 1)) & 
                (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
            ]
        
        if len(filtered_df) == 0:
            filtered_df = df_2024[df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05)]
        
        engine_power = filtered_df['Engine Power (hp)'].median()
        if pd.isna(engine_power):
            engine_power = 75  # Default placeholder if no data available
        return engine_power

    engine_power = calculate_engine_power(make, model, year, engine_size)
    
    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)
    
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)
    
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    user_data_final = user_data_final[required_columns]
    user_data_scaled = scaler.transform(user_data_final)

    model_predicted_price = model.predict(user_data_scaled)[0]

    def calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type):
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Transmission'] == transmission) & 
            (df_2024['Fuel Type'] == fuel_type) & 
            (df_2024['Body Type'] == body_type) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) & 
                (df_2024['Model'] == model) & 
                (df_2024['Year'] == year) & 
                (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
            ]

        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) & 
                (df_2024['Model'] == model) & 
                (df_2024['Year'] == year) & 
                (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
            ]

        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) & 
                (df_2024['Model'] == model) & 
                (df_2024['Year'].between(year - 1, year + 1)) & 
                (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
            ]

        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) & 
                (df_2024['Model'] == model) & 
                (df_2024['Year'].between(year - 2, year + 2)) & 
                (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
            ]

        if len(filtered_df) < 10:
            for i in range(1, 5):
                filtered_df = df_2024[
                    (df_2024['Make'] == make) &
                    (df_2024['Model'] == model) &
                    (df_2024['Year'].between(year - i, year + i)) &
                    (df_2024['Kilometers (km)'].between(km - 15000 * (i + 1), km + 15000 * (i + 1)))
                ]
                if len(filtered_df) >= 10:
                    break

        if len(filtered_df) < 10:
            filtered_df = df_2024[
                (df_2024['Make'] == make) &
                (df_2024['Model'] == model)
            ]

        if len(filtered_df) < 10:
            base_confidence = "Low"
        elif len(filtered_df) < 30:
            base_confidence = "Medium"
        else:
            base_confidence = "High"

        if len(filtered_df) >= 10:
            filtered_df = filtered_df.sort_values(by='Price USD')
            base_line_price = filtered_df['Price USD'].median()
        else:
            base_line_price = filtered_df['Price USD'].median() if len(filtered_df) > 0 else overall_median_price

        base_std_dev = filtered_df['Price USD'].std()

        return base_line_price, base_confidence, filtered_df, base_std_dev

    overall_median_price = df_2024['Price USD'].median()
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(
        make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type)

    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    def calculate_weights(base_confidence, model_confidence, agreement_level):
        if base_confidence == "High" and model_confidence == "High":
            if agreement_level == "High":
                return 0.6, 0.4
            elif agreement_level == "Medium":
                return 0.65, 0.35
            else:
                return 0.7, 0.3
        elif base_confidence == "High" and model_confidence == "Low":
            if agreement_level == "High":
                return 0.7, 0.3
            elif agreement_level == "Medium":
                return 0.75, 0.25
            else:
                return 0.8, 0.2
        elif base_confidence == "Low" and model_confidence == "High":
            if agreement_level == "High":
                return 0.5, 0.5
            elif agreement_level == "Medium":
                return 0.55, 0.45
            else:
                return 0.6, 0.4
        else:
            if agreement_level == "High":
                return 0.55, 0.45
            elif agreement_level == "Medium":
                return 0.6, 0.4
            else:
                return 0.65, 0.35

    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    result = f"""
    Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, and {traction} traction: {final_price:.2f} USD
    Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})
    Model confidence level: {model_confidence} - {"Model confidence is low due to the model's test max error being " + f"{ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}." if model_confidence == "Low" else f"Model confidence is high due to the model's test max error being {ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}."}
    Agreement level: {agreement_level} due to percentage difference being {percentage_difference:.2f}%.
    Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%
    """
    return result

with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=makes)
    model = gr.Dropdown(label="Model")
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Kilometers (km)", precision=0)
    color = gr.Dropdown(label="Color", choices=colors)
    transmission = gr.Dropdown(label="Transmission")
    body_type = gr.Dropdown(label="Body Type")
    fuel_type = gr.Dropdown(label="Fuel Type")
    traction = gr.Dropdown(label="Traction")
    estimate_button = gr.Button("Estimate Car Value")
    estimated_value = gr.Textbox(label="Estimated Car Value", interactive=False)

    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, transmission, body_type, fuel_type], outputs=traction)
    estimate_button.click(fn=estimate_car_value, inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction], outputs=estimated_value)

demo.launch()
```

```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
import gradio as gr
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

# Define dropdown choices
makes = sorted(df_2024['Make'].unique())
colors = sorted(df_2024['Color'].unique())

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.update(choices=models, value=models[0] if models else None)

def update_transmissions(make, model, year):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year)]
    transmissions = sorted(filtered_df['Transmission'].unique())
    return gr.update(choices=transmissions, value=transmissions[0] if transmissions else None)

def update_body_types(make, model, year, transmission):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & (df_2024['Transmission'] == transmission)]
    body_types = sorted(filtered_df['Body Type'].unique())
    return gr.update(choices=body_types, value=body_types[0] if body_types else None)

def update_fuel_types(make, model, year, transmission, body_type):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & 
                          (df_2024['Transmission'] == transmission) & (df_2024['Body Type'] == body_type)]
    fuel_types = sorted(filtered_df['Fuel Type'].unique())
    return gr.update(choices=fuel_types, value=fuel_types[0] if fuel_types else None)

def update_tractions(make, model, year, transmission, body_type, fuel_type):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & 
                          (df_2024['Transmission'] == transmission) & (df_2024['Body Type'] == body_type) & 
                          (df_2024['Fuel Type'] == fuel_type)]
    tractions = sorted(filtered_df['Traction'].unique())
    return gr.update(choices=tractions, value=tractions[0] if tractions else None)

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()
    make_model = f"{make}_{model}".replace(' ', '_')

    model_path = f"MakeModel/{make_model}_model.pkl"
    scaler_path = f"MakeModel/{make_model}_scaler.pkl"
    encoder_path = "MakeModel/onehot_encoder.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return f"Model not available for {make_model}. Unable to estimate price."

    car_age = datetime.now().year - year
    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type)

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)

    # Load the encoder and scaler
    encoder = joblib.load(encoder_path)
    scaler = joblib.load(scaler_path)

    # Specify the numerical and categorical columns
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    # Required columns by the model
    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    # Encode the categorical features
    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))

    # Concatenate the numerical features and encoded categorical features
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    # Ensure all required columns are present
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    # Reorder the columns to match the model's expected input
    user_data_final = user_data_final[required_columns]

    # Scale the user data
    user_data_scaled = scaler.transform(user_data_final)

    # Load the model
    model = joblib.load(model_path)

    # Predict the price using the model
    model_predicted_price = model.predict(user_data_scaled)[0]

    # Print the base price and the model predicted price
    base_line_price = float(base_line_price)  # Convert to float for rounding
    model_predicted_price = float(model_predicted_price)  # Convert to float for rounding

    # Calculate agreement level
    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    # Calculate confidence in the model's prediction using test max error percentage and value count
    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    # Determine weights based on confidences and agreement
    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)

    # Calculate the final price estimate
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    return f"Estimated Price: {final_price:.2f} USD"

def calculate_engine_power(make, model, year, engine_size):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'].between(year - 1, year + 1)) & 
        (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
    ]
    
    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
        
    if len(filtered_df) == 0:
        filtered_df = df_2024[
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
    
    if len(filtered_df) == 0:
        filtered_df = df_2024[df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05)]
    
    engine_power = filtered_df['Engine Power (hp)'].median()
    if pd.isna(engine_power):
        engine_power = 75  # Default placeholder if no data available
    return engine_power

def calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Fuel Type'] == fuel_type) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
    ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 2, year + 2)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        for i in range(1, 5):
            filtered_df = df_2024[
                (df_2024['Make'] == make) &
                (df_2024['Model'] == model) &
                (df_2024['Year'].between(year - i, year + i)) &
                (df_2024['Kilometers (km)'].between(km - 15000 * (i + 1), km + 15000 * (i + 1)))
            ]
            if len(filtered_df) >= 10:
                break

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) &
            (df_2024['Model'] == model)
        ]

    if len(filtered_df) < 10:
        base_confidence = "Low"
    elif len(filtered_df) < 30:
        base_confidence = "Medium"
    else:
        base_confidence = "High"

    if len(filtered_df) >= 10:
        filtered_df = filtered_df.sort_values(by='Price USD')
        base_line_price = filtered_df['Price USD'].median()
    else:
        base_line_price = filtered_df['Price USD'].median() if len(filtered_df) > 0 else overall_median_price

    base_std_dev = filtered_df['Price USD'].std()

    return base_line_price, base_confidence, filtered_df, base_std_dev

def calculate_weights(base_confidence, model_confidence, agreement_level):
    if base_confidence == "High" and model_confidence == "High":
        if agreement_level == "High":
            return 0.6, 0.4
        elif agreement_level == "Medium":
            return 0.65, 0.35
        else:
            return 0.7, 0.3
    elif base_confidence == "High" and model_confidence == "Low":
        if agreement_level == "High":
            return 0.7, 0.3
        elif agreement_level == "Medium":
            return 0.75, 0.25
        else:
            return 0.8, 0.2
    elif base_confidence == "Low" and model_confidence == "High":
        if agreement_level == "High":
            return 0.5, 0.5
        elif agreement_level == "Medium":
            return 0.55, 0.45
        else:
            return 0.6, 0.4
    else:
        if agreement_level == "High":
            return 0.55, 0.45
        elif agreement_level == "Medium":
            return 0.6, 0.4
        else:
            return 0.65, 0.35

with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=makes)
    model = gr.Dropdown(label="Model")
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Km", precision=0)
    color = gr.Dropdown(label="Color", choices=colors)
    transmission = gr.Dropdown(label="Transmission")
    body_type = gr.Dropdown(label="Body Type")
    fuel_type = gr.Dropdown(label="Fuel Type")
    traction = gr.Dropdown(label="Traction")
    estimate_button = gr.Button("Estimate Car Value")
    output = gr.Textbox(label="Estimated car value")

    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, transmission, body_type, fuel_type], outputs=traction)
    estimate_button.click(fn=estimate_car_value, inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction], outputs=output)

demo.launch()
```

```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
import gradio as gr
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

makes = sorted(df_2024['Make'].unique())
colors = sorted(df_2024['Color'].unique())

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.Dropdown.update(choices=models)

def update_transmissions(make, model, year):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year)
    ]
    transmissions = sorted(filtered_df['Transmission'].unique())
    return gr.Dropdown.update(choices=transmissions)

def update_body_types(make, model, year, transmission):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission)
    ]
    body_types = sorted(filtered_df['Body Type'].unique())
    return gr.Dropdown.update(choices=body_types)

def update_fuel_types(make, model, year, transmission, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type)
    ]
    fuel_types = sorted(filtered_df['Fuel Type'].unique())
    return gr.Dropdown.update(choices=fuel_types)

def update_tractions(make, model, year, transmission, body_type, fuel_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Fuel Type'] == fuel_type)
    ]
    tractions = sorted(filtered_df['Traction'].unique())
    return gr.Dropdown.update(choices=tractions)

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()
    make_model = f"{make}_{model}".replace(' ', '_')

    model_path = f"/mnt/data/MakeModel/{make_model}_model.pkl"
    scaler_path = f"/mnt/data/MakeModel/{make_model}_scaler.pkl"
    encoder_path = "/mnt/data/MakeModel/onehot_encoder.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return f"Model not available for {make_model}. Unable to estimate price."

    car_age = datetime.now().year - year

    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type)

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)

    # Load the encoder and scaler
    encoder = joblib.load(encoder_path)
    scaler = joblib.load(scaler_path)

    # Specify the numerical and categorical columns
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    # Required columns by the model
    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    # Encode the categorical features
    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))

    # Concatenate the numerical features and encoded categorical features
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    # Ensure all required columns are present
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    # Reorder the columns to match the model's expected input
    user_data_final = user_data_final[required_columns]

    # Scale the user data
    user_data_scaled = scaler.transform(user_data_final)

    # Load the model
    model = joblib.load(model_path)

    # Predict the price using the model
    model_predicted_price = model.predict(user_data_scaled)[0]

    # Print the base price and the model predicted price
    base_line_estimate = f"Base-line estimate: {base_line_price:.2f} USD"
    model_predicted_estimate = f"Model predicted price: {model_predicted_price:.2f} USD"

    # Calculate agreement level
    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    # Calculate confidence in the model's prediction using test max error percentage and value count
    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    # Determine weights based on confidences and agreement
    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)

    # Calculate the final price estimate
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    # Print the final estimated price and confidence levels
    final_estimate = f"Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, {traction} traction: {final_price:.2f} USD"
    base_confidence_level = f"Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})"
    
    model_confidence_reason = (
        f"Model confidence is low due to the model's test max error being {ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}."
        if model_confidence == "Low"
        else f"Model confidence is high due to the model's test max error being {ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}."
    )
    model_confidence_level = f"Model confidence level: {model_confidence} - {model_confidence_reason}"

    agreement_level_reason = f"Agreement level: {agreement_level} due to percentage difference being {percentage_difference:.2f}%."
    weights_info = f"Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%"
    
    filtered_data_info = "Filtered data used for base-line estimate:\n" + base_filtered_data[['Price USD', 'Year', 'Kilometers (km)', 'Engine Size (cc)', 'Body Type', 'Transmission']].to_string(index=False)

    non_zero_data = user_data_final.loc[:, (user_data_final != 0).any(axis=0)]
    input_data_info = "Input data for model prediction (non-zero values):\n" + non_zero_data.to_string(index=False)

    return (
        final_estimate, base_line_estimate, model_predicted_estimate, base_confidence_level,
        model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info
    )

# Gradio interface
with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=makes)
    model = gr.Dropdown(label="Model")
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Kilometers (km)", precision=0)
    color = gr.Dropdown(label="Color", choices=colors)
    transmission = gr.Dropdown(label="Transmission")
    body_type = gr.Dropdown(label="Body Type")
    fuel_type = gr.Dropdown(label="Fuel Type")
    traction = gr.Dropdown(label="Traction")
    
    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, transmission, body_type, fuel_type], outputs=traction)
    
    estimate_button = gr.Button("Estimate Car Value")
    
    final_estimate_output = gr.Textbox(label="Estimated Car Value")
    base_line_estimate_output = gr.Textbox(label="Base-line Estimate")
    model_predicted_estimate_output = gr.Textbox(label="Model Predicted Estimate")
    base_confidence_level_output = gr.Textbox(label="Base-line Confidence Level")
    model_confidence_level_output = gr.Textbox(label="Model Confidence Level")
    agreement_level_reason_output = gr.Textbox(label="Agreement Level Reason")
    weights_info_output = gr.Textbox(label="Weights Info")
    filtered_data_info_output = gr.Textbox(label="Filtered Data Info")
    input_data_info_output = gr.Textbox(label="Input Data Info")
    
    estimate_button.click(
        fn=estimate_car_value,
        inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction],
        outputs=[
            final_estimate_output, base_line_estimate_output, model_predicted_estimate_output,
            base_confidence_level_output, model_confidence_level_output, agreement_level_reason_output,
            weights_info_output, filtered_data_info_output, input_data_info_output
        ]
    )

demo.launch()
```

```python
import gradio as gr
import pandas as pd
import joblib
import numpy as np
from datetime import datetime
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

# Load pre-trained model and encoders (assuming they exist in the specified path)
def load_model_and_encoders(make_model):
    model_path = f"MakeModel/{make_model}_model.pkl"
    scaler_path = f"MakeModel/{make_model}_scaler.pkl"
    encoder_path = "MakeModel/onehot_encoder.pkl"
    model = joblib.load(model_path)
    scaler = joblib.load(scaler_path)
    encoder = joblib.load(encoder_path)
    return model, scaler, encoder

def get_unique_values(column):
    return df_2024[column].unique().tolist()

makes = sorted(get_unique_values('Make'))
colors = sorted(get_unique_values('Color'))

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.update(choices=models, value=models[0] if models else None)

def update_transmissions(make, model, year):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year)]
    transmissions = filtered_df['Transmission'].value_counts().index.tolist()
    return gr.update(choices=transmissions, value=transmissions[0] if transmissions else None)

def update_body_types(make, model, year, transmission):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & (df_2024['Transmission'] == transmission)]
    body_types = filtered_df['Body Type'].value_counts().index.tolist()
    return gr.update(choices=body_types, value=body_types[0] if body_types else None)

def update_fuel_types(make, model, year, transmission, body_type):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & (df_2024['Transmission'] == transmission) & (df_2024['Body Type'] == body_type)]
    fuel_types = filtered_df['Fuel Type'].value_counts().index.tolist()
    return gr.update(choices=fuel_types, value=fuel_types[0] if fuel_types else None)

def update_tractions(make, model, year, transmission, body_type, fuel_type):
    filtered_df = df_2024[(df_2024['Make'] == make) & (df_2024['Model'] == model) & (df_2024['Year'] == year) & (df_2024['Transmission'] == transmission) & (df_2024['Body Type'] == body_type) & (df_2024['Fuel Type'] == fuel_type)]
    tractions = filtered_df['Traction'].value_counts().index.tolist()
    return gr.update(choices=tractions, value=tractions[0] if tractions else None)

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()

    car_age = datetime.now().year - year
    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(
        make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type
    )

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }

    user_data_df = pd.DataFrame(preset_data)
    encoder = joblib.load("MakeModel/onehot_encoder.pkl")
    scaler = joblib.load(f"MakeModel/{make}_{model}_scaler.pkl")
    model = joblib.load(f"MakeModel/{make}_{model}_model.pkl")

    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)',
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric',
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline',
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van',
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3',
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van',
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster',
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black',
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold',
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray',
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink',
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray',
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4',
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    user_data_final = user_data_final[required_columns]
    user_data_scaled = scaler.transform(user_data_final)
    model_predicted_price = model.predict(user_data_scaled)[0]

    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    agreement_level = "High" if percentage_difference <= 5 else "Medium" if percentage_difference <= 10 else "Low"

    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    final_estimate = f"Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, {ownership} ownership, and {traction} traction: {final_price:.2f} USD"
    base_confidence_level = f"Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})"

    if model_confidence == "Low":
        model_confidence_reason = f"Model confidence is low due to the model's test max error being {ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}."
    else:
        model_confidence_reason = f"Model confidence is high due to the model's test max error being {ml_test_max_error_percent:.2f}% and the value count being {ml_test_value_count}."

    model_confidence_level = f"Model confidence level: {model_confidence} - {model_confidence_reason}"
    agreement_level_reason = f"Agreement level: {agreement_level} due to percentage difference being {percentage_difference:.2f}%."
    weights_info = f"Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%"

    filtered_data_info = "Filtered data used for base-line estimate:\n" + base_filtered_data[['Price USD', 'Year', 'Kilometers (km)', 'Engine Size (cc)', 'Body Type', 'Transmission']].to_string(index=False)
    non_zero_data = user_data_final.loc[:, (user_data_final != 0).any(axis=0)]
    input_data_info = "Input data for model prediction (non-zero values):\n" + non_zero_data.to_string(index=False)

    return final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info

with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=makes)
    model = gr.Dropdown(label="Model")
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Kilometers (km)", precision=0)
    color = gr.Dropdown(label="Color", choices=colors)
    transmission = gr.Dropdown(label="Transmission")
    body_type = gr.Dropdown(label="Body Type")
    fuel_type = gr.Dropdown(label="Fuel Type")
    traction = gr.Dropdown(label="Traction")
    output = gr.Textbox(label="Estimated Car Value")

    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    year.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, transmission, body_type, fuel_type], outputs=traction)

    def on_submit(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
        final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info = estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction)
        return f"{final_estimate}\n\n{base_confidence_level}\n{model_confidence_level}\n{agreement_level_reason}\n{weights_info}\n{filtered_data_info}\n\n{input_data_info}"

    submit_btn = gr.Button("Estimate Car Value")
    submit_btn.click(fn=on_submit, inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction], outputs=output)

demo.launch()
```

```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
import gradio as gr
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

def match_user_input(user_input, choices, cutoff=60):
    if not user_input:
        return None
    user_input = user_input.strip().title()  # Standardize user input
    match = process.extractOne(user_input, choices, scorer=fuzz.WRatio, score_cutoff=cutoff)
    return match[0] if match else None

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.Dropdown.update(choices=models, value=models[0] if models else None)

def update_transmissions(make, model, year):
    filtered_df = df_2024[
        (df_2024['Make'] == make) &
        (df_2024['Model'] == model) &
        (df_2024['Year'] == year)
    ]
    transmissions = filtered_df['Transmission'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=transmissions, value=transmissions[0] if transmissions else None)

def update_body_types(make, model, year, transmission):
    filtered_df = df_2024[
        (df_2024['Make'] == make) &
        (df_2024['Model'] == model) &
        (df_2024['Year'] == year) &
        (df_2024['Transmission'] == transmission)
    ]
    body_types = filtered_df['Body Type'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=body_types, value=body_types[0] if body_types else None)

def update_fuel_types(make, model, year, transmission, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) &
        (df_2024['Model'] == model) &
        (df_2024['Year'] == year) &
        (df_2024['Transmission'] == transmission) &
        (df_2024['Body Type'] == body_type)
    ]
    fuel_types = filtered_df['Fuel Type'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=fuel_types, value=fuel_types[0] if fuel_types else None)

def update_tractions(make, model, year, transmission, body_type, fuel_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) &
        (df_2024['Model'] == model) &
        (df_2024['Year'] == year) &
        (df_2024['Transmission'] == transmission) &
        (df_2024['Body Type'] == body_type) &
        (df_2024['Fuel Type'] == fuel_type)
    ]
    tractions = filtered_df['Traction'].value_counts().index.tolist()
    return gr.Dropdown.update(choices=tractions, value=tractions[0] if tractions else None)

def calculate_engine_power(make, model, year, engine_size):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'].between(year - 1, year + 1)) & 
        (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
    ]
    
    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
        
    if len(filtered_df) == 0:
        filtered_df = df_2024[
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05))
        ]
    
    if len(filtered_df) == 0:
        filtered_df = df_2024[df_2024['Engine Size (cc)'].between(engine_size * 0.95, engine_size * 1.05)]
    
    engine_power = filtered_df['Engine Power (hp)'].median()
    if pd.isna(engine_power):
        engine_power = 75  # Default placeholder if no data available
    return engine_power

def calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type):
    filtered_df = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Fuel Type'] == fuel_type) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
    ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'] == year) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 1, year + 1)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) & 
            (df_2024['Model'] == model) & 
            (df_2024['Year'].between(year - 2, year + 2)) & 
            (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
        ]

    if len(filtered_df) < 10:
        for i in range(1, 5):
            filtered_df = df_2024[
                (df_2024['Make'] == make) &
                (df_2024['Model'] == model) &
                (df_2024['Year'].between(year - i, year + i)) &
                (df_2024['Kilometers (km)'].between(km - 15000 * (i + 1), km + 15000 * (i + 1)))
            ]
            if len(filtered_df) >= 10:
                break

    if len(filtered_df) < 10:
        filtered_df = df_2024[
            (df_2024['Make'] == make) &
            (df_2024['Model'] == model)
        ]

    if len(filtered_df) < 10:
        base_confidence = "Low"
    elif len(filtered_df) < 30:
        base_confidence = "Medium"
    else:
        base_confidence = "High"

    if len(filtered_df) >= 10:
        filtered_df = filtered_df.sort_values(by='Price USD')
        base_line_price = filtered_df['Price USD'].median()
    else:
        base_line_price = filtered_df['Price USD'].median() if len(filtered_df) > 0 else overall_median_price

    base_std_dev = filtered_df['Price USD'].std()

    return base_line_price, base_confidence, filtered_df, base_std_dev

def calculate_weights(base_confidence, model_confidence, agreement_level):
    if base_confidence == "High" and model_confidence == "High":
        if agreement_level == "High":
            return 0.6, 0.4
        elif agreement_level == "Medium":
            return 0.65, 0.35
        else:
            return 0.7, 0.3
    elif base_confidence == "High" and model_confidence == "Low":
        if agreement_level == "High":
            return 0.7, 0.3
        elif agreement_level == "Medium":
            return 0.75, 0.25
        else:
            return 0.8, 0.2
    elif base_confidence == "Low" and model_confidence == "High":
        if agreement_level == "High":
            return 0.5, 0.5
        elif agreement_level == "Medium":
            return 0.55, 0.45
        else:
            return 0.6, 0.4
    else:
        if agreement_level == "High":
            return 0.55, 0.45
        elif agreement_level == "Medium":
            return 0.6, 0.4
        else:
            return 0.65, 0.35

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()
    car_age = datetime.now().year - year
    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(
        make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type)

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)

    make_model = f"{make}_{model}".replace(' ', '_')
    model_path = f"MakeModel/{make_model}_model.pkl"
    scaler_path = f"MakeModel/{make_model}_scaler.pkl"
    encoder_path = "MakeModel/onehot_encoder.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return f"Model not available for {make_model}. Unable to estimate price."

    # Load the encoder and scaler
    encoder = joblib.load(encoder_path)
    scaler = joblib.load(scaler_path)

    # Specify the numerical and categorical columns
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    # Required columns by the model
    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    # Encode the categorical features
    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))

    # Concatenate the numerical features and encoded categorical features
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    # Ensure all required columns are present
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    # Reorder the columns to match the model's expected input
    user_data_final = user_data_final[required_columns]

    # Scale the user data
    user_data_scaled = scaler.transform(user_data_final)

    # Load the model
    model = joblib.load(model_path)

    # Predict the price using the model
    model_predicted_price = model.predict(user_data_scaled)[0]

    # Print the base price and the model predicted price
    print(f"Base-line estimate: {base_line_price:.2f} USD")
    print(f"Model predicted price: {model_predicted_price:.2f} USD")

    # Calculate agreement level
    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    # Calculate confidence in the model's prediction using test max error percentage and value count
    ml_test_max_error = base_filtered_data.iloc[0]['ML Test Max Error']
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    # Determine weights based on confidences and agreement
    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)

    # Calculate the final price estimate
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    # Print the final estimated price and confidence levels
    final_estimate = f"Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, Owner ownership, and {traction} traction: {final_price:.2f} USD"
    base_confidence_level = f"Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})"
    
    model_confidence_level = f"Model confidence level: {model_confidence} - {'Model confidence is low due to the model test max error being ' + str(ml_test_max_error_percent) + '% and the value count being ' + str(ml_test_value_count) if model_confidence == 'Low' else 'Model confidence is high due to the model test max error being ' + str(ml_test_max_error_percent) + '% and the value count being ' + str(ml_test_value_count)}"
    agreement_level_reason = f"Agreement level: {agreement_level} due to percentage difference being {percentage_difference:.2f}%."
    weights_info = f"Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%"

    filtered_data_info = "Filtered data used for base-line estimate:\n" + base_filtered_data[['Price USD', 'Year', 'Kilometers (km)', 'Engine Size (cc)', 'Body Type', 'Transmission']].to_string(index=False)
    non_zero_data = user_data_final.loc[:, (user_data_final != 0).any(axis=0)]
    input_data_info = "Input data for model prediction (non-zero values):\n" + non_zero_data.to_string(index=False)

    return final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info

# Set up Gradio interface
def on_submit(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info = estimate_car_value(
        make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction)
    return final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info

with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=sorted(df_2024['Make'].unique()))
    model = gr.Dropdown(label="Model")
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Kilometers (km)", precision=0)
    color = gr.Dropdown(label="Color", choices=sorted(df_2024['Color'].unique()))
    transmission = gr.Dropdown(label="Transmission")
    body_type = gr.Dropdown(label="Body Type")
    fuel_type = gr.Dropdown(label="Fuel Type")
    traction = gr.Dropdown(label="Traction")
    
    final_estimate = gr.Textbox(label="Estimated Car Value")
    base_confidence_level = gr.Textbox(label="Base-line Confidence Level")
    model_confidence_level = gr.Textbox(label="Model Confidence Level")
    agreement_level_reason = gr.Textbox(label="Agreement Level")
    weights_info = gr.Textbox(label="Weights Info")
    filtered_data_info = gr.Textbox(label="Filtered Data Info")
    input_data_info = gr.Textbox(label="Input Data Info")

    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    year.change(fn=update_transmissions, inputs=[make, model, year], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, transmission, body_type, fuel_type], outputs=traction)

    submit_button = gr.Button("Estimate Car Value")
    submit_button.click(fn=on_submit, inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction],
                        outputs=[final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info])

demo.launch()
```

```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime
import os
import gradio as gr
from rapidfuzz import process, fuzz

# Define the filename of your CSV file
clean_csv_file = '2024_turkey_car_market_ML.csv'

# Load the data from the CSV file into a new DataFrame
df_2024 = pd.read_csv(clean_csv_file)

# Disable scientific notation for printing
pd.set_option('display.float_format', lambda x: '%.6f' % x)
np.set_printoptions(suppress=True)

# Create 'Make_Model' column
df_2024['Make_Model'] = df_2024['Make'] + ' ' + df_2024['Model']

def match_user_input(user_input, choices, cutoff=60):
    if not user_input:
        return None
    user_input = user_input.strip().title()  # Standardize user input
    match = process.extractOne(user_input, choices, scorer=fuzz.WRatio, score_cutoff=cutoff)
    print(f"Matching '{user_input}' against choices: {match}")  # Debugging print statement
    return match[0] if match else None

def update_models(make):
    models = sorted(df_2024[df_2024['Make'] == make]['Model'].unique())
    return gr.update(choices=models, value=models[0] if models else None)

def update_transmissions(make, model, year, km):
    filtered_transmissions = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3))
    ]['Transmission'].value_counts().index.tolist()
    return gr.update(choices=filtered_transmissions, value=filtered_transmissions[0] if filtered_transmissions else None)

def update_body_types(make, model, year, km, transmission):
    filtered_body_types = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3)) & 
        (df_2024['Transmission'] == transmission)
    ]['Body Type'].value_counts().index.tolist()
    return gr.update(choices=filtered_body_types, value=filtered_body_types[0] if filtered_body_types else None)

def update_fuel_types(make, model, year, km, transmission, body_type):
    filtered_fuel_types = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3)) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type)
    ]['Fuel Type'].value_counts().index.tolist()
    return gr.update(choices=filtered_fuel_types, value=filtered_fuel_types[0] if filtered_fuel_types else None)

def update_tractions(make, model, year, km, transmission, body_type, fuel_type):
    filtered_tractions = df_2024[
        (df_2024['Make'] == make) & 
        (df_2024['Model'] == model) & 
        (df_2024['Year'] == year) & 
        (df_2024['Kilometers (km)'].between(km * 0.7, km * 1.3)) & 
        (df_2024['Transmission'] == transmission) & 
        (df_2024['Body Type'] == body_type) & 
        (df_2024['Fuel Type'] == fuel_type)
    ]['Traction'].value_counts().index.tolist()
    return gr.update(choices=filtered_tractions, value=filtered_tractions[0] if filtered_tractions else None)

def estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    overall_median_price = df_2024['Price USD'].median()
    car_age = datetime.now().year - year
    engine_power = calculate_engine_power(make, model, year, engine_size)
    base_line_price, base_confidence, base_filtered_data, base_std_dev = calculate_base_price(make, model, year, km, engine_size, overall_median_price, transmission, fuel_type, body_type)

    preset_data = {
        'Kilometers (km)': [km],
        'Car Age (Years)': [car_age],
        'Engine Size (cc)': [engine_size],
        'Engine Power (hp)': [engine_power],
        'Transmission': [transmission],
        'Fuel Type': [fuel_type],
        'Body Type': [body_type],
        'Color': [color],
        'Traction': [traction]
    }
    
    user_data_df = pd.DataFrame(preset_data)

    # Load the encoder and scaler
    encoder_path = "MakeModel/onehot_encoder.pkl"
    encoder = joblib.load(encoder_path)
    model_path = f"MakeModel/{make}_{model.replace(' ', '_')}_model.pkl"
    scaler_path = f"MakeModel/{make}_{model.replace(' ', '_')}_scaler.pkl"
    
    if not os.path.exists(model_path) or not os.path.exists(scaler_path) or not os.path.exists(encoder_path):
        return "Model not available. Unable to estimate price."

    scaler = joblib.load(scaler_path)

    # Specify the numerical and categorical columns
    numerical_cols = ['Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)']
    categorical_cols = ['Transmission', 'Fuel Type', 'Body Type', 'Color', 'Traction']

    # Required columns by the model
    required_columns = [
        'Kilometers (km)', 'Car Age (Years)', 'Engine Size (cc)', 'Engine Power (hp)', 
        'Transmission_Manual', 'Transmission_Semi-Automatic', 'Fuel Type_Electric', 
        'Fuel Type_Gasoline', 'Fuel Type_Hybrid', 'Fuel Type_LPG and Gasoline', 
        'Body Type_Cabrio', 'Body Type_Coupe', 'Body Type_Crossover', 'Body Type_Glass Van', 
        'Body Type_Half Glass Van', 'Body Type_Hard top', 'Body Type_Hatchback/3', 
        'Body Type_Hatchback/5', 'Body Type_MPV', 'Body Type_Minibus', 'Body Type_Panel Van', 
        'Body Type_Pick-up', 'Body Type_Refrigerated Panel Van', 'Body Type_Roadster', 
        'Body Type_SUV', 'Body Type_Sedan', 'Body Type_Station wagon', 'Color_Black', 
        'Color_Blue', 'Color_Brown', 'Color_Champagne', 'Color_Dark Red', 'Color_Gold', 
        'Color_Gray', 'Color_Green', 'Color_Metallic Blue', 'Color_Metallic Gray', 
        'Color_Metallic Green', 'Color_Navy Blue', 'Color_Orange', 'Color_Pink', 
        'Color_Purple', 'Color_Red', 'Color_Silver', 'Color_Smoke', 'Color_Titanium Gray', 
        'Color_Turquoise', 'Color_White', 'Color_Yellow', 'Traction_4x4', 
        'Traction_AWD (Electronic)', 'Traction_Front-wheel-drive', 'Traction_Rear-wheel-drive'
    ]

    # Encode the categorical features
    user_data_encoded = pd.DataFrame(encoder.transform(user_data_df[categorical_cols]), columns=encoder.get_feature_names_out(categorical_cols))

    # Concatenate the numerical features and encoded categorical features
    user_data_final = pd.concat([user_data_df[numerical_cols], user_data_encoded], axis=1)

    # Ensure all required columns are present
    for col in required_columns:
        if col not in user_data_final.columns:
            user_data_final[col] = 0

    # Reorder the columns to match the model's expected input
    user_data_final = user_data_final[required_columns]

    # Scale the user data
    user_data_scaled = scaler.transform(user_data_final)

    # Load the model
    model = joblib.load(model_path)

    # Predict the price using the model
    model_predicted_price = model.predict(user_data_scaled)[0]

    # Calculate agreement level
    percentage_difference = abs(model_predicted_price - base_line_price) / base_line_price * 100
    if percentage_difference <= 5:
        agreement_level = "High"
    elif percentage_difference <= 10:
        agreement_level = "Medium"
    else:
        agreement_level = "Low"

    # Calculate confidence in the model's prediction using test max error percentage and value count
    ml_test_value_count = base_filtered_data.iloc[0]['ML MakeModel value count']
    ml_test_max_error_percent = (ml_test_max_error / base_line_price) * 100

    model_confidence = "High" if ml_test_value_count >= 30 and ml_test_max_error_percent <= 10 else "Low"

    # Determine weights based on confidences and agreement
    base_weight, model_weight = calculate_weights(base_confidence, model_confidence, agreement_level)

    # Calculate the final price estimate
    final_price = (base_weight * base_line_price) + (model_weight * model_predicted_price)

    # Print the final estimated price and confidence levels
    final_estimate = f"Estimated Price for {make} {model} {year} with {km} km, {color} color, {transmission} transmission, {body_type} body type, {fuel_type} fuel type, {engine_size} cc engine, Owner ownership, and {traction} traction: {final_price:.2f} USD"
    base_confidence_level = f"Base-line confidence level: {base_confidence} (Standard Deviation: {base_std_dev:.2f})"
    
    model_confidence_level = f"Model confidence level: {model_confidence} - {'Model confidence is low due to the models test max error being ' + str(ml_test_max_error_percent) + '% and the value count being ' + str(ml_test_value_count) if model_confidence == 'Low' else 'Model confidence is high due to the models test max error being ' + str(ml_test_max_error_percent) + '% and the value count being ' + str(ml_test_value_count)}"
    
    agreement_level_reason = f"Agreement level: {agreement_level} due to percentage difference being {percentage_difference:.2f}%."
    weights_info = f"Weights: Base-line {base_weight*100:.1f}%, Model {model_weight*100:.1f}%"
    
    filtered_data_info = "Filtered data used for base-line estimate:\n" + base_filtered_data[['Price USD', 'Year', 'Kilometers (km)', 'Engine Size (cc)', 'Body Type', 'Transmission']].to_string(index=False)
    non_zero_data = user_data_final.loc[:, (user_data_final != 0).any(axis=0)]
    input_data_info = "Input data for model prediction (non-zero values):\n" + non_zero_data.to_string(index=False)

    return final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info

# Set up Gradio interface
def on_submit(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction):
    final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info = estimate_car_value(make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction)
    return final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info

with gr.Blocks() as demo:
    make = gr.Dropdown(label="Make", choices=sorted(df_2024['Make'].unique()))
    model = gr.Dropdown(label="Model", interactive=True)
    year = gr.Number(label="Year", precision=0)
    engine_size = gr.Number(label="Engine Size (cc)", precision=0)
    km = gr.Number(label="Kilometers (km)", precision=0)
    color = gr.Dropdown(label="Color", choices=sorted(df_2024['Color'].unique()))
    transmission = gr.Dropdown(label="Transmission", interactive=True)
    body_type = gr.Dropdown(label="Body Type", interactive=True)
    fuel_type = gr.Dropdown(label="Fuel Type", interactive=True)
    traction = gr.Dropdown(label="Traction", interactive=True)
    estimate_button = gr.Button("Estimate Car Value")
    final_estimate = gr.Textbox(label="Estimated car value")
    base_confidence_level = gr.Textbox(label="Base-line confidence level")
    model_confidence_level = gr.Textbox(label="Model confidence level")
    agreement_level_reason = gr.Textbox(label="Agreement level")
    weights_info = gr.Textbox(label="Weights")
    filtered_data_info = gr.Textbox(label="Filtered data")
    input_data_info = gr.Textbox(label="Input data for model prediction")

    make.change(fn=update_models, inputs=make, outputs=model)
    model.change(fn=update_transmissions, inputs=[make, model, year, km], outputs=transmission)
    transmission.change(fn=update_body_types, inputs=[make, model, year, km, transmission], outputs=body_type)
    body_type.change(fn=update_fuel_types, inputs=[make, model, year, km, transmission, body_type], outputs=fuel_type)
    fuel_type.change(fn=update_tractions, inputs=[make, model, year, km, transmission, body_type, fuel_type], outputs=traction)
    estimate_button.click(fn=on_submit, inputs=[make, model, year, engine_size, km, color, transmission, body_type, fuel_type, traction], 
                         outputs=[final_estimate, base_confidence_level, model_confidence_level, agreement_level_reason, weights_info, filtered_data_info, input_data_info])

demo.launch()
```
